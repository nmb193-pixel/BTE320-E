```python
sales = float(input("Enter Sales: "))
cogs = sales * 0.40
gross_profit = sales - cogs
tax_rate = float(input("Enter Tax Rate (%): "))
tax_amount = gross_profit * (tax_rate / 100)
net_profit = gross_profit - tax_amount
print(f"Net Sales: ${sales}")
print(f"Cost of Goods Sold: ${cogs}")
print(f"Gross Profit: ${gross_profit}")
print(f"Tax Amount: ${tax_amount}")
print(f"Net Profit: ${net_profit}")
```

```python
number1 = float(input("Number 1: "))
sign = input ("Operator: ")
number2 = float(input("Number 2: "))
if sign == "+":
  print(number1 + number2)
elif sign == "-":
  print(number1 - number2)
elif sign == "*":
  print(number1 * number2)
elif sign == "/":
    print(number1 / number2)
elif sign == "^":
      print(number1 ** number2)
elif sign == "%":
        print(number1 % number2)
else:
    print("Invalid Operator")
```

```python
base = 5
rate = 2.5
min_fee = 10
max_fee = 20

hours = 1
while hours <= 8:
    fee = base + (rate * hours)

    if fee < min_fee:
        fee = min_fee
    if fee > max_fee:
        fee = max_fee

    print(hours, fee)
    hours += 1
```

```python
ticket_price = 10
base_people = 20
fixed_cost = 200

print("Ad$ Profit")

ads = 0
while ads <= 200:
    extra_people = 2 * (ads ** 0.5)
    total_people = base_people + extra_people

    revenue = total_people * ticket_price
    profit = revenue - fixed_cost - ads

    print(ads, round(profit, 1))

    ads += 25
```

```python
import random

play = input("Do you want to play Rock, Paper, Scissors? (yes/no): ")

while play == "yes":

    # user move
    user_move = input("Enter rock, paper, or scissors: ")

    # computer randomly chooses
    computer_move = random.choice(["rock", "paper", "scissors"])

    print("Computer chose:", computer_move)

    if user_move == computer_move:
        print("Tie!")

    elif user_move == "rock":
        if computer_move == "scissors":
            print("You win!")
        else:
            print("Computer wins!")

    elif user_move == "paper":
        if computer_move == "rock":
            print("You win!")
        else:
            print("Computer wins!")

    elif user_move == "scissors":
        if computer_move == "paper":
            print("You win!")
        else:
            print("Computer wins!")

    else:
        print("Invalid input.")

    # ask to play again
    play = input("Do you want to play again? (yes/no): ")

print("Game over.")
```

```python
name_counts = {}
N = 5

i = 0
while i < N:
    name = input(f"Enter name {i+1}/{N}: ")
    if name in name_counts:
        name_counts[name] += 1
    else:
        name_counts[name] = 1
    i += 1

print("Final Name Counts:")
print(name_counts)
```

```python
def parkingGarage(hours):
  money = 5 + 2.5 * hours
  if money < 10.0:
    money = 10.0
  elif money > 20:
    money = 20.0
  print(f"You pay ${money} for {hours} hours")
  print("--------------------------")
  return money

print("Parking Garage Prices\n")
for i in range(1, 9):
  parkingGarage(i)
```

```python
def reverse(s):
    if len(s) == 0:
        return s
    else:
        return reverse(s[1:]) + s[0]

s = input('')
print(reverse(s))
```

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def display(self, formal: bool):
        if formal:
            print(f"{self.last_name}, {self.first_name}")
        else:
            print(self.first_name)

    def __str__(self):
        return f"{self.first_name} {self.last_name}"


# Example usage
p = Person("John", "Doe")

p.display(formal=True)   # Doe, John
p.display(formal=False)  # John
print(p)                 # John Doe
```
